# Acceleration of a Body in Circular Orbit

We derive the acceleration for a particle moving in a circle of radius $r$ with angular position $\theta(t)$.

For uniform circular motion, $r$ is constant and $\dot{\theta} = \omega$ is also constant. The result will be the familiar centripetal acceleration directed toward the center.

## 1. Position Vector

Write the position vector in Cartesian coordinates as

$$
\vec r(t) = r\cos\theta(t)\,\hat i + r\sin\theta(t)\,\hat j
$$

Since the body remains on a circle, $r$ is constant. Differentiate with respect to time to get the velocity:

$$
\vec v(t) = \frac{d\vec r}{dt}
= -r\dot\theta\sin\theta\,\hat i + r\dot\theta\cos\theta\,\hat j
$$

For uniform circular motion, $\dot\theta = \omega$, so

$$
\vec v(t) = -r\omega\sin\theta\,\hat i + r\omega\cos\theta\,\hat j
$$

A quick check that $\vec v$ is perpendicular to $\vec r$ is to compute the dot product:

$$
\vec r \cdot \vec v
= (r\cos\theta)(-r\omega\sin\theta) + (r\sin\theta)(r\omega\cos\theta) = 0
$$

Since the dot product is zero, $\vec r \perp \vec v$. Geometrically, $\vec r$ points radially outward and $\vec v$ is tangent to the circle.

It is often clearer to introduce the radial and tangential unit vectors

$$
\hat r = \cos\theta\,\hat i + \sin\theta\,\hat j,
\qquad
\hat\theta = -\sin\theta\,\hat i + \cos\theta\,\hat j
$$

Then

$$
\vec r = r\hat r,
\qquad
\vec v = r\dot\theta\,\hat\theta
$$

This makes the roles of the quantities clear: $\dot\theta$ is a scalar angular speed, while $\hat\theta$ is the tangential direction. Since $\hat r \perp \hat\theta$, the velocity is perpendicular to the radius.

### Why is $v = \omega r$?

This is the relation between the **speed** and the angular speed.

If the particle moves through an angle $\theta$, the distance traveled along the circle is the arc length

$$
s = r\theta
$$

Differentiate with respect to time:

$$
v = \frac{ds}{dt} = r\frac{d\theta}{dt} = r\omega
$$

So:

- $v$ is the linear speed along the tangent to the circle.
- $\omega$ is the angular speed in radians per second.
- Multiplying angular speed by radius converts rotation rate into ordinary linear speed.

You can also see this from the velocity vector above. Its magnitude is

$$
|\vec v| = \sqrt{(-r\omega\sin\theta)^2 + (r\omega\cos\theta)^2}
= r\omega\sqrt{\sin^2\theta + \cos^2\theta}
= r\omega
$$

So $v = r\omega$ is the magnitude of the tangential velocity, not a new assumption.

## 1.5 Visualizing $\vec r$, $\vec v$, and $\vec a$

In the diagram below:

- $\vec r$ points from the center of the circle to the particle. This is the **radial outward** direction.
- $\vec v$ is tangent to the circle, so it is perpendicular to $\vec r$. Its direction is the direction in which $\theta$ is increasing.
- $\vec a$ points inward toward the center for uniform circular motion. This is the **radial inward** or centripetal direction.

Using polar unit vectors,

$$
\vec r = r\hat r, \qquad \vec v = r\dot\theta\,\hat\theta
$$

and, more generally,

$$
\vec a = -r\dot\theta^2\hat r + r\ddot\theta\,\hat\theta.
$$

For **uniform** circular motion, $r$ is constant and $\ddot\theta = 0$, so

$$
\vec a = -r\dot\theta^2\hat r = -\omega^2 \vec r.
$$

A useful way to remember the directions is:

- $\hat r$ points outward from the center.
- $\hat\theta$ is perpendicular to $\hat r$ and points in the direction of increasing angle $\theta$.
- So if the motion is counterclockwise, $\vec v$ points counterclockwise along the tangent.
- If the motion is clockwise, then $\dot\theta < 0$, so $\vec v$ points the opposite way along the tangent.
- But $\vec a$ depends on $\dot\theta^2$, not on the sign of $\dot\theta$, so for uniform circular motion it still points inward even when the motion is clockwise.

So the acceleration would point outward only if some different force law were acting. In ordinary circular motion, the inward direction is what keeps the particle on the circle.

This makes the formulas

$$
\vec v \perp \vec r, \qquad \vec a = -\omega^2 \vec r
$$

much easier to interpret geometrically.

The plotted velocity arrows in the figure correspond to **increasing** $\theta$, so they show the counterclockwise case. If you reverse the direction of motion, the green $\vec v$ arrows flip, but the red $\vec a$ arrows still point toward the center.

```python
import numpy as np
import matplotlib.pyplot as plt

radius = 3.0
sample_angles = [0, np.pi / 2, np.pi, 3 * np.pi / 2]
panel_titles = [r'$\theta = 0$', r'$\theta = \pi/2$', r'$\theta = \pi$', r'$\theta = 3\pi/2$']

fig, axes = plt.subplots(2, 2, figsize=(10, 10))
angles = np.linspace(0, 2 * np.pi, 400)

for ax, theta0, title in zip(axes.flat, sample_angles, panel_titles):
    x = radius * np.cos(theta0)
    y = radius * np.sin(theta0)

    ax.plot(radius * np.cos(angles), radius * np.sin(angles), label='Orbit')

    ax.arrow(0, 0, x, y, head_width=0.18, length_includes_head=True, color='tab:blue')
    ax.text(x / 2 + 0.12, y / 2 + 0.12, r'$\vec r$', color='tab:blue', fontsize=12)

    vx = -np.sin(theta0)
    vy = np.cos(theta0)
    ax.arrow(x, y, vx, vy, head_width=0.18, length_includes_head=True, color='tab:green')
    ax.text(x + vx + 0.12, y + vy + 0.12, r'$\vec v$', color='tab:green', fontsize=12)

    ax.arrow(x, y, -0.9 * x, -0.9 * y, head_width=0.18, length_includes_head=True, color='tab:red')
    ax.text(x - 0.55 * x - 0.25, y - 0.55 * y - 0.1, r'$\vec a$', color='tab:red', fontsize=12)

    ax.scatter([0, x], [0, y], color=['black', 'tab:orange'])
    ax.text(x + 0.12, y - 0.28, 'particle', fontsize=10)
    ax.text(0.1, -0.28, 'center', fontsize=10)

    ax.set_aspect('equal')
    ax.set_xlim(-4, 4)
    ax.set_ylim(-4, 4)
    ax.axhline(0, color='gray', linewidth=0.8)
    ax.axvline(0, color='gray', linewidth=0.8)
    ax.set_title(title)

fig.suptitle('Directions of $\\vec r$, $\\vec v$, and $\\vec a$ in Circular Motion', fontsize=14)
handles, labels = axes.flat[0].get_legend_handles_labels()
fig.legend(handles, labels, loc='upper right')
plt.tight_layout()
plt.show()
```

## 1.75 Alternate Derivation of $\left|\vec a\right| = \frac{v^2}{r}$

This derivation is based on the **isosceles triangle** formed by two nearby velocity vectors.

Take the velocity at two nearby instants, $\vec v_1$ and $\vec v_2$. In uniform circular motion:

- $|\vec v_1| = |\vec v_2| = v$, because the speed is constant.
- The radius vector turns by a small angle $\Delta\theta$ between the two instants.
- At every point, the velocity is tangent to the circle, so it is perpendicular to the radius.
- Rotating both radius vectors by the same fixed angle $90^\circ$ gives the two velocity directions.

A rotation by the same fixed angle does not change the angle between two vectors. Therefore the angle between $\vec v_1$ and $\vec v_2$ is also $\Delta\theta$.

When the two vectors are drawn from the same tail, they form an isosceles triangle. The third side is

$$
\Delta \vec v = \vec v_2 - \vec v_1
$$

Using the law of cosines on that triangle,

$$
|\Delta \vec v|^2 = v^2 + v^2 - 2v^2\cos(\Delta\theta)
$$

so

$$
|\Delta \vec v| = \sqrt{2v^2(1-\cos\Delta\theta)} = 2v\sin\left(\frac{\Delta\theta}{2}\right)
$$

For a very small angle, $\sin(x) \approx x$, hence

$$
|\Delta \vec v| \approx 2v\left(\frac{\Delta\theta}{2}\right) = v\Delta\theta
$$

Now divide by $\Delta t$:

$$
\frac{|\Delta \vec v|}{\Delta t} \approx v\frac{\Delta\theta}{\Delta t}
$$

Taking the limit as $\Delta t \to 0$,

$$
|\vec a| = \lim_{\Delta t \to 0} \frac{|\Delta \vec v|}{\Delta t} = v\frac{d\theta}{dt} = v\omega
$$

Using $v = r\omega$,

$$
|\vec a| = v\left(\frac{v}{r}\right) = \frac{v^2}{r}
$$

Therefore,

$$
\boxed{|\vec a| = \frac{v^2}{r}}
$$

and the direction is inward because, for nearby points, $\Delta\vec v$ points toward the center.

## 1.8 Alternate Derivation Using the Complex Plane

The complex-plane viewpoint is useful because direction is built in. A point in the plane can be written as

$$
z(t) = x(t) + iy(t).
$$

For motion on a circle of fixed radius $r$,

$$
z(t) = r e^{i\theta(t)}.
$$

This is the same as

$$
x(t) = r\cos\theta(t), \qquad y(t) = r\sin\theta(t),
$$

but the factor $e^{i\theta}$ keeps track of the direction automatically.

A key fact is that multiplying by $i$ rotates a vector by $90^\circ$ counterclockwise. So if $z$ points radially outward, then $iz$ points in the tangential direction.

Differentiate once:

$$
\dot z = r \frac{d}{dt} \left(e^{i\theta}\right) = r i\dot\theta e^{i\theta} = i\dot\theta z.
$$

So the velocity is

$$
\dot z = i\dot\theta z.
$$

This already shows two things:

- $\dot z$ is obtained by rotating $z$ by $90^\circ$, so velocity is tangent to the circle.
- The sign of $\dot\theta$ determines clockwise versus counterclockwise motion.

For uniform circular motion, $\dot\theta = \omega$ is constant, so

$$
\dot z = i\omega z.
$$

Differentiate again:

$$
\ddot z = \frac{d}{dt}(i\omega z) = i\omega \dot z.
$$

Now substitute $\dot z = i\omega z$:

$$
\ddot z = i\omega (i\omega z) = -\omega^2 z.
$$

So the acceleration is

$$
\ddot z = -\omega^2 z.
$$

This is the complex-plane version of

$$
\vec a = -\omega^2 \vec r.
$$

The minus sign means the acceleration points in the direction opposite to $z$, that is, opposite to the outward radial vector. Therefore it points inward toward the center.

This also makes the clockwise case clear. If the motion is clockwise, then $\omega < 0$, so the velocity changes sign because

$$
\dot z = i\omega z,
$$

but the acceleration does not flip outward because

$$
\ddot z = -\omega^2 z
$$

depends on $\omega^2$, which is always positive.

Finally, taking magnitudes gives

$$
|\ddot z| = \omega^2 |z| = \omega^2 r.
$$

Since $v = |\dot z| = |\omega|r$, we get

$$
|\vec a| = \omega^2 r = \frac{v^2}{r}.
$$

So the complex-plane method gives both the magnitude and the inward direction in a very compact way.

### Sketch for the Alternate Derivation

The right-hand figure is the isosceles triangle used in the derivation.

- $\vec v_1$ and $\vec v_2$ start from the same point and have equal length $v$.
- The angle between them is $\Delta\theta$ because each velocity is obtained by rotating the corresponding radius by $90^\circ$.
- Rotating both vectors by the same amount preserves the angle between them.
- The side joining their tips has length $|\Delta\vec v|$.

For small $\Delta\theta$, that base points almost directly inward, which is why the acceleration is centripetal.

```python
import numpy as np
import matplotlib.pyplot as plt

radius = 3.0
theta1 = np.pi / 4
delta_theta = 0.35
theta2 = theta1 + delta_theta
speed = 1.2

p1 = np.array([radius * np.cos(theta1), radius * np.sin(theta1)])
p2 = np.array([radius * np.cos(theta2), radius * np.sin(theta2)])

v1 = speed * np.array([-np.sin(theta1), np.cos(theta1)])
v2 = speed * np.array([-np.sin(theta2), np.cos(theta2)])
delta_v = v2 - v1

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

angle = np.linspace(0, 2 * np.pi, 400)
axes[0].plot(radius * np.cos(angle), radius * np.sin(angle), color='0.5', linestyle='--')
axes[0].arrow(0, 0, p1[0], p1[1], head_width=0.14, length_includes_head=True, color='tab:blue')
axes[0].arrow(0, 0, p2[0], p2[1], head_width=0.14, length_includes_head=True, color='tab:cyan')
axes[0].arrow(p1[0], p1[1], v1[0], v1[1], head_width=0.12, length_includes_head=True, color='tab:green')
axes[0].arrow(p2[0], p2[1], v2[0], v2[1], head_width=0.12, length_includes_head=True, color='tab:olive')
axes[0].text(p1[0] + 0.1, p1[1] + 0.1, r'$P_1$')
axes[0].text(p2[0] + 0.1, p2[1] + 0.1, r'$P_2$')
axes[0].text(p1[0] + v1[0] + 0.05, p1[1] + v1[1], r'$\vec v_1$', color='tab:green')
axes[0].text(p2[0] + v2[0] + 0.05, p2[1] + v2[1], r'$\vec v_2$', color='tab:olive')
axes[0].text(0.55, 2.9, r'$\Delta\theta$')
axes[0].set_title('Nearby Velocities on the Orbit')
axes[0].set_aspect('equal')
axes[0].set_xlim(-4, 4)
axes[0].set_ylim(-4, 4)
axes[0].axhline(0, color='0.85', linewidth=0.8)
axes[0].axvline(0, color='0.85', linewidth=0.8)

origin = np.array([0.0, 0.0])
tip1 = v1
tip2 = v2
triangle_x = [origin[0], tip1[0], tip2[0], origin[0]]
triangle_y = [origin[1], tip1[1], tip2[1], origin[1]]

axes[1].fill(triangle_x, triangle_y, color='tab:red', alpha=0.08, zorder=1)
axes[1].plot([origin[0], tip1[0]], [origin[1], tip1[1]], color='tab:green', linewidth=2.5, zorder=2)
axes[1].plot([origin[0], tip2[0]], [origin[1], tip2[1]], color='tab:olive', linewidth=2.5, zorder=2)
axes[1].plot([tip1[0], tip2[0]], [tip1[1], tip2[1]], color='tab:red', linewidth=2.5, zorder=2)

axes[1].arrow(origin[0], origin[1], tip1[0], tip1[1], head_width=0.05, length_includes_head=True, color='tab:green', zorder=3)
axes[1].arrow(origin[0], origin[1], tip2[0], tip2[1], head_width=0.05, length_includes_head=True, color='tab:olive', zorder=3)

axes[1].scatter([origin[0], tip1[0], tip2[0]], [origin[1], tip1[1], tip2[1]], color='black', s=28, zorder=4)
axes[1].text(origin[0] + 0.03, origin[1] + 0.03, 'O')

midpoint = (tip1 + tip2) / 2
axes[1].text(tip1[0] + 0.04, tip1[1] + 0.04, r'$\vec v_1$', color='tab:green')
axes[1].text(tip2[0] - 0.10, tip2[1] - 0.02, r'$\vec v_2$', color='tab:olive')
axes[1].text(midpoint[0] - 0.03, midpoint[1] + 0.04, r'$|\Delta \vec v|$', color='tab:red')

angle1 = np.arctan2(tip1[1], tip1[0])
angle2 = np.arctan2(tip2[1], tip2[0])
arc_angles = np.linspace(angle2, angle1, 50)
arc_radius = 0.18
axes[1].plot(arc_radius * np.cos(arc_angles), arc_radius * np.sin(arc_angles), color='black', linewidth=1.2)
arc_mid = (angle1 + angle2) / 2
axes[1].text(0.24 * np.cos(arc_mid), 0.24 * np.sin(arc_mid) + 0.02, r'$\Delta\theta$')

axes[1].text(midpoint[0] - 0.10, midpoint[1] - 0.14, r'$\Delta \vec v = \vec v_2 - \vec v_1$', color='tab:red')

axes[1].set_title('Isosceles Velocity Triangle')
axes[1].set_aspect('equal')
axes[1].set_xlim(-1.35, 0.25)
axes[1].set_ylim(-0.1, 1.1)
axes[1].axhline(0, color='0.9', linewidth=0.8)
axes[1].axvline(0, color='0.9', linewidth=0.8)

plt.tight_layout()
plt.show()
```

## 2. Differentiate Again for Acceleration

Differentiate the velocity once more:

$$
\vec a(t) = \frac{d\vec v}{dt}
$$

Using the product rule,

$$
\vec a(t) = -r\left(\ddot\theta\sin\theta + \dot\theta^2\cos\theta\right)\hat i
\;+\; r\left(\ddot\theta\cos\theta - \dot\theta^2\sin\theta\right)\hat j
$$

For uniform circular motion, $\ddot\theta = 0$ and $\dot\theta = \omega$, hence

$$
\vec a(t) = -r\omega^2\cos\theta\,\hat i - r\omega^2\sin\theta\,\hat j
$$

Factor out $-\omega^2$:

$$
\vec a(t) = -\omega^2\left(r\cos\theta\,\hat i + r\sin\theta\,\hat j\right)
= -\omega^2\vec r(t)
$$

Since $v = r\omega$, this can also be written as

$$
\boxed{\vec a = -\frac{v^2}{r}\,\hat r}
$$

So the acceleration has magnitude

$$
|\vec a| = r\omega^2 = \frac{v^2}{r}
$$

and always points toward the center of the circle. This is the centripetal acceleration.

```python
import sympy as sp

t, r, omega = sp.symbols('t r omega', positive=True, real=True)
theta = omega * t

position = sp.Matrix([r * sp.cos(theta), r * sp.sin(theta)])
velocity = sp.diff(position, t)
acceleration = sp.diff(velocity, t)

print('position =')
sp.pprint(position)
print('\nvelocity =')
sp.pprint(sp.simplify(velocity))
print('\nacceleration =')
sp.pprint(sp.simplify(acceleration))
print('\nacceleration + omega^2 * position =')
sp.pprint(sp.simplify(acceleration + omega**2 * position))
```
