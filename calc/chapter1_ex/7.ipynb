A man standing on a wharf is hauling a rope attached to a boat at the rate of four feet a second. If his handes are nine feet above the point of attachment , how fast is the boat approaching the wharf when it is 12 feet away.

## Solution

Let

- $x(t)$ be the horizontal distance of the boat from the wharf, in feet
- $s(t)$ be the length of the rope, in feet

The man's hands are 9 ft above the point of attachment, so the rope, the vertical segment, and the horizontal distance form a right triangle:

$$
s^2 = x^2 + 9^2 = x^2 + 81
$$

The rope is being hauled in at 4 ft/s, so its length is decreasing:

$$
\frac{ds}{dt} = -4 \text{ ft/s}
$$

Differentiate the relation with respect to time:

$$
2s\frac{ds}{dt} = 2x\frac{dx}{dt}
$$

So

$$
s\frac{ds}{dt} = x\frac{dx}{dt}
$$

When the boat is 12 ft from the wharf,

$$
x = 12
$$

and from the triangle,

$$
s = \sqrt{12^2 + 9^2} = \sqrt{144+81} = \sqrt{225} = 15
$$

Substitute into

$$
s\frac{ds}{dt} = x\frac{dx}{dt}
$$

$$
15(-4) = 12\frac{dx}{dt}
$$

$$
-60 = 12\frac{dx}{dt}
$$

$$
\frac{dx}{dt} = -5 \text{ ft/s}
$$

## Answer

The negative sign means the distance from the wharf is decreasing, so the boat is approaching the wharf at

$$
\boxed{5 \text{ ft/s}}
$$
