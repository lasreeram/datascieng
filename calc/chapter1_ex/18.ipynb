The hypotenuse $AB$ of a right angle triangle $ABC$ remains 5 inches long while the other two sides change, the side $AC$ increasing at the rate of 2 in/min. At what rate is the area of the triangle changing when $AC$ is just 3 inches?

## Solution

Let

- $x(t) = AC$
- $y(t) = BC$

Since the hypotenuse $AB$ is constant at 5 inches,

$$x^2 + y^2 = 25$$

We are given

$$\frac{dx}{dt} = 2 \text{ in/min}$$

and we want $\dfrac{dA}{dt}$ when $x = 3$.

The area of the triangle is

$$A = \frac{1}{2}xy$$

Differentiate with respect to time:

$$\frac{dA}{dt} = \frac{1}{2}\left(x\frac{dy}{dt} + y\frac{dx}{dt}\right)$$

So we need $\dfrac{dy}{dt}$.

Differentiate the constraint equation:

$$x^2 + y^2 = 25$$
$$2x\frac{dx}{dt} + 2y\frac{dy}{dt} = 0$$

Hence

$$x\frac{dx}{dt} + y\frac{dy}{dt} = 0$$
$$\frac{dy}{dt} = -\frac{x}{y}\frac{dx}{dt}$$

When $x=3$,

$$y = \sqrt{25-9} = 4$$

Thus

$$\frac{dy}{dt} = -\frac{3}{4}(2) = -\frac{3}{2} \text{ in/min}$$

Now substitute into the area formula:

$$\frac{dA}{dt} = \frac{1}{2}\left(3\left(-\frac{3}{2}\right) + 4(2)\right)$$

$$\frac{dA}{dt} = \frac{1}{2}\left(-\frac{9}{2} + 8\right)$$

$$\frac{dA}{dt} = \frac{1}{2}\left(\frac{7}{2}\right) = \frac{7}{4}$$

## Answer

The area is increasing at

$$\boxed{\frac{7}{4} \text{ in}^2/\text{min}}$$

that is,

$$\boxed{1.75 \text{ in}^2/\text{min}}$$

## Alternative Method: Write $A$ Directly as a Function of $t$

Let $t=0$ at the instant when $AC=3$.

Since $AC$ is increasing at 2 in/min,

$$x(t) = 3 + 2t$$

Because the hypotenuse is always 5,

$$y(t) = \sqrt{25 - x(t)^2} = \sqrt{25 - (3+2t)^2}$$

Therefore the area is

$$A(t) = \frac{1}{2}x(t)y(t) = \frac{1}{2}(3+2t)\sqrt{25-(3+2t)^2}$$

Differentiate using the product rule:

$$\frac{dA}{dt} = \frac{1}{2}\left[2\sqrt{25-(3+2t)^2} + (3+2t)\frac{d}{dt}\left(\sqrt{25-(3+2t)^2}\right)\right]$$

Now use the chain rule:

$$\frac{d}{dt}\left(\sqrt{25-(3+2t)^2}\right) = \frac{1}{2\sqrt{25-(3+2t)^2}}\cdot\left(-2(3+2t)(2)\right)$$

so

$$\frac{d}{dt}\left(\sqrt{25-(3+2t)^2}\right) = -\frac{2(3+2t)}{\sqrt{25-(3+2t)^2}}$$

Substitute back:

$$\frac{dA}{dt} = \frac{1}{2}\left[2\sqrt{25-(3+2t)^2} - \frac{2(3+2t)^2}{\sqrt{25-(3+2t)^2}}\right]$$

At $t=0$,

$$\sqrt{25-3^2} = \sqrt{16} = 4$$

Hence

$$\left.\frac{dA}{dt}\right|_{t=0} = \frac{1}{2}\left[2(4) - \frac{2(9)}{4}\right]$$

$$\left.\frac{dA}{dt}\right|_{t=0} = \frac{1}{2}\left[8 - \frac{9}{2}\right] = \frac{1}{2}\cdot\frac{7}{2} = \frac{7}{4}$$

So this method gives the same result:

$$\boxed{\frac{dA}{dt} = \frac{7}{4} \text{ in}^2/\text{min}}$$

```python
import math

x = 3
dx_dt = 2
y = math.sqrt(25 - x**2)
dy_dt = -(x / y) * dx_dt
dA_dt = 0.5 * (x * dy_dt + y * dx_dt)

print(f"y = {y}")
print(f"dy/dt = {dy_dt} in/min")
print(f"dA/dt = {dA_dt} in^2/min")
print("Exact value = 7/4 in^2/min")
```
