A metal plate in the shape of an equilateral triangle is being heated in such a way that each of the sides is increasing at the rate of ten inches per hour. How rapidly is the area increasing at the instant when each side is 69.28 inches.

## Solution

Let $s(t)$ be the side length of the equilateral triangle, in inches. We are given

$$
\frac{ds}{dt} = 10 \text{ in/hr}
$$

The area of an equilateral triangle with side length $s$ is

$$
A = \frac{\sqrt{3}}{4}s^2
$$

Differentiate with respect to time:

$$
\frac{dA}{dt} = \frac{\sqrt{3}}{4} \cdot 2s\frac{ds}{dt}
$$

So

$$
\frac{dA}{dt} = \frac{\sqrt{3}}{2}s\frac{ds}{dt}
$$

At the instant when

$$
s = 69.28 \text{ in}, \qquad \frac{ds}{dt} = 10 \text{ in/hr}
$$

we get

$$
\frac{dA}{dt} = \frac{\sqrt{3}}{2}(69.28)(10)
$$

$$
\frac{dA}{dt} = 346.4\sqrt{3} \text{ in}^2/\text{hr}
$$

Using $\sqrt{3} \approx 1.73205$,

$$
\frac{dA}{dt} \approx 599.98 \text{ in}^2/\text{hr}
$$

## Answer

The area is increasing at approximately

$$
\boxed{600 \text{ in}^2/\text{hr}}
$$
