A block of building stone is to be lifted by rope 50 ft long passing over a pulley on a window ledge 25 feet above the level ground. A man takes hold of the loose end of the rope which is held five feet above the ground and walks away from the block at 10 feet a second. How rapidly will the block begin to rise?

## Solution

Let

- $x(t)$ be the horizontal distance from the wall to the man's hand
- $y(t)$ be the height of the stone above the ground

The pulley is 25 ft above the ground, and the man's hand is 5 ft above the ground, so the vertical distance from the hand to the pulley is

$$
25 - 5 = 20 \text{ ft}
$$

So the rope segment from the pulley to the man's hand has length

$$
\sqrt{x^2 + 20^2} = \sqrt{x^2 + 400}
$$

The other rope segment runs from the pulley down to the stone. Since the stone is $y$ feet above the ground, that segment has length

$$
25 - y
$$

Because the total rope length is 50 ft,

$$
(25 - y) + \sqrt{x^2 + 400} = 50
$$

Differentiate with respect to time:

$$
-\frac{dy}{dt} + \frac{x}{\sqrt{x^2+400}}\frac{dx}{dt} = 0
$$

So

$$
\frac{dy}{dt} = \frac{x}{\sqrt{x^2+400}}\frac{dx}{dt}
$$

The man walks away at

$$
\frac{dx}{dt} = 10 \text{ ft/s}
$$

At the instant he begins to walk away, he is directly below the pulley, so

$$
x = 0
$$

Therefore

$$
\frac{dy}{dt} = \frac{0}{\sqrt{400}} \cdot 10 = 0
$$

## Answer

The block begins to rise at

$$
\boxed{0 \text{ ft/s}}
$$

So at the exact instant the man starts walking away, the stone has zero upward velocity. It only starts moving upward after that instant.
