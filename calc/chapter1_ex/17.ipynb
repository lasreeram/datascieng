A tank is in the form of a cone with the point downward and the height and diameter are each 10 feet. How fast is the water pouring in at the moment when it is 5 feet deep and the surface is rising at the rate of 4 feet per minute?

## Solution

Let $h$ be the depth of the water and $r$ the radius of the water surface.

The full cone has:

- height $= 10$ ft
- radius $= 5$ ft

By similar triangles,

$$\frac{r}{h} = \frac{5}{10} = \frac{1}{2}$$

so

$$r = \frac{h}{2}$$

The volume of water in a cone is

$$V = \frac{1}{3}\pi r^2 h$$

Substitute $r = \dfrac{h}{2}$:

$$V = \frac{1}{3}\pi \left(\frac{h}{2}\right)^2 h$$
$$V = \frac{1}{3}\pi \cdot \frac{h^2}{4} \cdot h$$
$$V = \frac{\pi}{12} h^3$$

Differentiate with respect to time $t$:

$$\frac{dV}{dt} = \frac{\pi}{12} \cdot 3h^2 \frac{dh}{dt}$$
$$\frac{dV}{dt} = \frac{\pi}{4} h^2 \frac{dh}{dt}$$

Now substitute:

- $h = 5$ ft
- $\dfrac{dh}{dt} = 4$ ft/min

$$\frac{dV}{dt} = \frac{\pi}{4}(5^2)(4)$$
$$\frac{dV}{dt} = \frac{\pi}{4}(25)(4)$$
$$\frac{dV}{dt} = 25\pi$$

## Answer

The water is pouring in at

$$\boxed{25\pi \text{ ft}^3/\text{min}}$$

which is approximately

$$\boxed{78.54 \text{ ft}^3/\text{min}}$$

```python
import math

h = 5
dh_dt = 4

dV_dt = (math.pi / 4) * h**2 * dh_dt
print(f"dV/dt = {dV_dt:.6f} ft^3/min")
print("Exact value = 25*pi ft^3/min")
```
