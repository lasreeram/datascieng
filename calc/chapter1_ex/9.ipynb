A 3 mile wind blowing on a level is carrying a kite directly away from a boy. How high is the kite when it is directly over a point 100 feet away and he is paying out the string at the rate of 88ft a minute

## Solution

Let

- $x(t)$ = horizontal distance from the boy to the point directly under the kite, in feet
- $y(t)$ = height of the kite, in feet
- $s(t)$ = length of the string, in feet

Then

$$
s^2 = x^2 + y^2
$$

The wind is blowing **on a level**, so it carries the kite horizontally. That means at the instant in question the height is not changing, so

$$
\frac{dy}{dt} = 0
$$

Also, the wind speed is 3 miles per hour, which is

$$
3 \cdot 5280 = 15840 \text{ ft/hr} = 264 \text{ ft/min}
$$

So

$$
\frac{dx}{dt} = 264 \text{ ft/min}
$$

The boy is paying out the string at

$$
\frac{ds}{dt} = 88 \text{ ft/min}
$$

Differentiate

$$
s^2 = x^2 + y^2
$$

with respect to time:

$$
2s\frac{ds}{dt} = 2x\frac{dx}{dt} + 2y\frac{dy}{dt}
$$

Since $\frac{dy}{dt}=0$, this becomes

$$
s\frac{ds}{dt} = x\frac{dx}{dt}
$$

At the instant described, $x=100$, so

$$
s(88) = 100(264)
$$

$$
s = \frac{100 \cdot 264}{88} = 300 \text{ ft}
$$

Now use the right triangle relation:

$$
300^2 = 100^2 + y^2
$$

$$
y^2 = 90000 - 10000 = 80000
$$

$$
y = \sqrt{80000} = 200\sqrt{2} \text{ ft}
$$

## Answer

The kite is

$$
\boxed{200\sqrt{2} \text{ ft}}
$$

above the ground, which is approximately

$$
\boxed{282.8 \text{ ft}}
$$
