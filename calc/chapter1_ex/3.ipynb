two points move, one on OX axis and one on OY axis , in such a way that in t minutes their distances from O are x=2t^2-6t, y=6t-9. When will they be nearest together

## Solution

Let the two moving points be

- $(x,0)$ on the $x$-axis, where $x=2t^2-6t$
- $(0,y)$ on the $y$-axis, where $y=6t-9$

The distance between them is

$$
d = \sqrt{x^2+y^2}
$$

To find when they are nearest together, it is easier to minimize $d^2$:

$$
d^2 = x^2 + y^2 = (2t^2-6t)^2 + (6t-9)^2
$$

Expand:

$$
(2t^2-6t)^2 = 4t^4 - 24t^3 + 36t^2
$$

$$
(6t-9)^2 = 36t^2 - 108t + 81
$$

So

$$
d^2 = 4t^4 - 24t^3 + 72t^2 - 108t + 81
$$

Differentiate with respect to $t$:

$$
\frac{d}{dt}(d^2) = 16t^3 - 72t^2 + 144t - 108
$$

Factor out 4:

$$
16t^3 - 72t^2 + 144t - 108 = 4(4t^3 - 18t^2 + 36t - 27)
$$

Now factor the cubic correctly:

$$
4t^3 - 18t^2 + 36t - 27 = (2t-3)(2t^2-6t+9)
$$

since

$$
(2t-3)(2t^2-6t+9) = 4t^3 - 18t^2 + 36t - 27
$$

Therefore

$$
\frac{d}{dt}(d^2) = 4(2t-3)(2t^2-6t+9)
$$

Set this equal to zero:

$$
4(2t-3)(2t^2-6t+9)=0
$$

This gives

$$
2t-3=0 \quad \text{or} \quad 2t^2-6t+9=0
$$

The quadratic has no real roots, because

$$
(-6)^2 - 4(2)(9) = 36 - 72 = -36 < 0
$$

So the only real time is

$$
t=\frac{3}{2}
$$

Since $d^2$ decreases before $t=\frac{3}{2}$ and increases after, the distance is minimized there.

## Answer

The two points are nearest together when

$$
\boxed{t=\frac{3}{2}\text{ minutes}}
$$
