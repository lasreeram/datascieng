A rope 28 feet long is attached to a block on level ground and runs over a pulley 12 feet above the ground. The rope is stretched taut and the free end is drawn directly away from the block and pull at the rate of 13 ft per sec. How fast will the block be moving up when it is 5 feet away from the point directly below the pulley.

## Solution

Assume the motion is interpreted this way: the block rises vertically, and the free end of the rope is pulled horizontally away from the point directly below the pulley.

Let

- $x(t)$ be the horizontal distance of the free end from the point directly below the pulley.
- $y(t)$ be the height of the block above the ground.

Then the rope has two variable pieces:

- the vertical segment from the pulley down to the block has length $12-y$
- the slanted segment from the pulley to the free end has length $28-(12-y)=16+y$

That second identity is the key rope-length check: as the block rises, the left segment shortens by exactly the amount that the right segment lengthens.

The slanted segment, the ground, and the 12-ft pulley height form a right triangle, so

$$x^2 + 12^2 = (16+y)^2$$

That is,

$$x^2 + 144 = (16+y)^2$$

The free end is pulled away at 13 ft/sec, so

$$\frac{dx}{dt} = 13$$

We want $\dfrac{dy}{dt}$ when the block is 5 ft above the ground, so

$$y = 5$$

First solve for $x$ at that instant:

$$x^2 + 144 = (16+5)^2$$

$$x^2 + 144 = 441$$

$$x^2 = 297$$

$$x = \sqrt{297} = 3\sqrt{33}$$

Now differentiate

$$x^2 + 144 = (16+y)^2$$

with respect to $t$:

$$2x\frac{dx}{dt} = 2(16+y)\frac{dy}{dt}$$

So

$$x\frac{dx}{dt} = (16+y)\frac{dy}{dt}$$

and therefore

$$\frac{dy}{dt} = \frac{x}{16+y}\frac{dx}{dt}$$

Substitute $x = \sqrt{297}$, $y=5$, and $\dfrac{dx}{dt}=13$:

$$\frac{dy}{dt} = \frac{\sqrt{297}}{21}(13)$$

$$\frac{dy}{dt} = \frac{13\sqrt{297}}{21} = \frac{13\sqrt{33}}{7}$$

Numerically,

$$\frac{dy}{dt} \approx 10.67 \text{ ft/sec}$$

## Answer

Under this interpretation, the block is moving upward at

$$\boxed{\frac{13\sqrt{33}}{7} \text{ ft/sec} \approx 10.67 \text{ ft/sec}}$$

If the original exercise intended the block to stay on the ground instead, then $y$ would represent a different quantity and the model would change.

```python
import math

# Assumption used here: the block rises vertically and the free end is pulled horizontally.
# Total rope length is constant: (12 - y) + slanted_segment = 28.
# Therefore slanted_segment = 28 - (12 - y) = 16 + y, and

# x^2 + 12^2 = (16 + y)^2.

total_rope = 28
height_pulley = 12
y = 5
dx_dt = 13

vertical_segment = height_pulley - y
slanted_segment = total_rope - vertical_segment
x = math.sqrt(slanted_segment**2 - height_pulley**2)
dy_dt = (x / slanted_segment) * dx_dt

print(f"When y = {y} ft:")
print(f"vertical segment = 12 - y = {vertical_segment} ft")
print(f"slanted segment = 28 - (12 - y) = {slanted_segment} ft")
print(f"x = sqrt((16+y)^2 - 12^2) = {x:.6f} ft")
print(f"dy/dt = (x/(16+y)) dx/dt")
print(f"      = ({x:.6f}/{slanted_segment}) * {dx_dt}")
print(f"      = {dy_dt:.6f} ft/sec")
print(f"Exact form = 13*sqrt(33)/7 ft/sec")
```
