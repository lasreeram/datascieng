A man whose height is 6 feet walks directly from the lamp post at the rate of 3 miles/hr. on a level pavement. If the lamp is 10 feet above the pavment, at what rate is the end of the shadown travelling.

## Solution

Let

- $x(t)$ be the distance from the lamp post to the man
- $y(t)$ be the length of the man's shadow

Then the distance from the lamp post to the end of the shadow is

$$
x+y
$$

The man walks away from the lamp post at

$$
\frac{dx}{dt} = 3 \text{ miles/hr}
$$

Convert this to feet per hour:

$$
\frac{dx}{dt} = 3 \cdot 5280 = 15840 \text{ ft/hr}
$$

Using similar triangles:

- large triangle: height $10$, base $x+y$
- small triangle: height $6$, base $y$

So

$$
\frac{10}{x+y} = \frac{6}{y}
$$

Cross-multiply:

$$
10y = 6(x+y)
$$

$$
10y = 6x + 6y
$$

$$
4y = 6x
$$

$$
y = \frac{3}{2}x
$$

Differentiate with respect to time:

$$
\frac{dy}{dt} = \frac{3}{2}\frac{dx}{dt}
$$

So

$$
\frac{dy}{dt} = \frac{3}{2}(15840) = 23760 \text{ ft/hr}
$$

The end of the shadow is at distance $x+y$ from the lamp post, so its speed is

$$
\frac{d}{dt}(x+y) = \frac{dx}{dt} + \frac{dy}{dt}
$$

$$
= 15840 + 23760 = 39600 \text{ ft/hr}
$$

Convert to miles per hour:

$$
\frac{39600}{5280} = 7.5 \text{ miles/hr}
$$

## Answer

The end of the shadow is travelling at

$$
\boxed{7.5 \text{ miles/hr}}
$$
