# Beam Strength Problem

A beam of rectangular cross section has strength proportional to

$$
S \propto b d^2
$$

where:

- $b$ is the breadth
- $d$ is the depth

We want the **strongest** rectangular beam that can be cut from a round log of diameter 2 feet.

## Step 1: Geometry of the rectangle inside the log

A rectangle cut from a circular log is inscribed in the circle, so its diagonal equals the diameter of the log.

Since the diameter is 2 feet,

$$
b^2 + d^2 = 2^2 = 4.
$$

So the constraint is

$$
d^2 = 4 - b^2.
$$

## Step 2: Write strength as a function of one variable

Ignoring the constant of proportionality, maximize

$$
S = b d^2.
$$

Substitute $d^2 = 4 - b^2$:

$$
S(b) = b(4 - b^2) = 4b - b^3.
$$

## Step 3: Differentiate

$$
S'(b) = 4 - 3b^2.
$$

Set this equal to 0:

$$
4 - 3b^2 = 0
$$

$$
b^2 = \frac{4}{3}
$$

$$
b = \frac{2}{\sqrt{3}}.
$$

## Step 4: Find the corresponding depth

Using

$$
d^2 = 4 - b^2 = 4 - \frac{4}{3} = \frac{8}{3},
$$

we get

$$
d = \sqrt{\frac{8}{3}} = 2\sqrt{\frac{2}{3}}.
$$

## Final answer

The beam of greatest strength has:

$$
\boxed{b = \frac{2}{\sqrt{3}} \text{ ft}, \qquad d = 2\sqrt{\frac{2}{3}} \text{ ft}}
$$

and therefore

$$
\frac{d}{b} = \sqrt{2}.
$$

So the **shape** of the strongest beam is:

$$
\boxed{\text{depth} = \sqrt{2} \times \text{breadth}}
$$

That is, the depth should be $\sqrt{2}$ times the breadth.

```python
import sympy as sp

b = sp.symbols('b', positive=True)
S = 4*b - b**3
critical_b = sp.solve(sp.diff(S, b), b)
critical_b = [value for value in critical_b if value.is_real and value > 0][0]
critical_d = sp.sqrt(4 - critical_b**2)
ratio = sp.simplify(critical_d / critical_b)

print(f'breadth = {sp.simplify(critical_b)} ft')
print(f'depth = {sp.simplify(critical_d)} ft')
print(f'depth / breadth = {ratio}')
```

# Ship Fuel Economy Problem

The fuel cost per hour varies as the **cube of the speed**.

We are told:

- fuel cost is $25 dollars per hour when the speed is 10$ miles per hour
- all other expenses are a fixed $100 dollars per hour$

We want the **most economical speed**, which means the speed that minimizes the **total cost per mile**.

## Step 1: Write the fuel cost function

Let $v$ be the speed in miles per hour.

Since fuel cost varies as the cube of speed,

$$
F = kv^3
$$

for some constant $k$.

Given that $F = 25$ when $v = 10$,

$$
25 = k(10)^3 = 1000k
$$

so

$$
k = \frac{25}{1000} = \frac{1}{40}.
$$

Therefore the fuel cost per hour is

$$
F(v) = \frac{v^3}{40}.
$$

## Step 2: Write the total hourly cost

The other expenses are fixed at $100 dollars per hour$, so the total cost per hour is

$$
C_{\text{hour}}(v) = \frac{v^3}{40} + 100.
$$

## Step 3: Convert to cost per mile

At speed $v$ miles per hour, the ship travels $v$ miles in one hour.

So the cost per mile is

$$
C(v) = \frac{\frac{v^3}{40} + 100}{v} = \frac{v^2}{40} + \frac{100}{v}.
$$

This is the quantity we minimize.

## Intuition: why cost per mile?

If we only look at **cost per hour**, then there is no meaningful shipping objective built into the problem.

A ship is used to cover distance, so being economical means:

$$
\text{cost per distance} = \frac{\text{cost per hour}}{\text{distance per hour}}.
$$

That is exactly **cost per mile**.

Another way to see it is to imagine a trip of fixed length $D$ miles. Then

$$
\text{time} = \frac{D}{v}
$$

so the total cost of the trip is

$$
\left(\frac{v^3}{40} + 100\right)\frac{D}{v} = D\left(\frac{v^2}{40} + \frac{100}{v}\right).
$$

Since $D$ is fixed, minimizing the total trip cost is the same as minimizing

$$
\frac{v^2}{40} + \frac{100}{v},
$$

which is the cost per mile.

This also explains why cost per hour alone is not enough: going very slowly may reduce fuel burned per hour, but it also makes the trip take much longer.

## Step 4: Differentiate

$$
C'(v) = \frac{v}{20} - \frac{100}{v^2}.
$$

Set this equal to 0:

$$
\frac{v}{20} - \frac{100}{v^2} = 0.
$$

Multiply by $20v^2$:

$$
v^3 - 2000 = 0.
$$

So

$$
v^3 = 2000
$$

and hence

$$
v = \sqrt[3]{2000} = 10\sqrt[3]{2}.
$$

## Step 5: Confirm it is a minimum

$$
C''(v) = \frac{1}{20} + \frac{200}{v^3} > 0
$$

for positive $v$, so this critical point gives a minimum.

## Final answer

The most economical speed is

$$
\boxed{v = 10\sqrt[3]{2} \text{ miles per hour}}
$$

which is approximately

$$
\boxed{12.6 \text{ miles per hour}}.
$$

```python
v = sp.symbols('v', positive=True)
fuel_cost = v**3 / 40
total_cost_per_mile = sp.simplify((fuel_cost + 100) / v)
critical_v = sp.solve(sp.diff(total_cost_per_mile, v), v)
critical_v = [value for value in critical_v if value.is_real and value > 0][0]

print(f'cost per mile = {total_cost_per_mile}')
print(f'most economical speed = {sp.simplify(critical_v)} mph')
print(f'approximation = {sp.N(critical_v, 6)} mph')
```

# Telephone Company Revenue Problem

A telephone company charges:

- $20 per subscriber$ if there are 100 subscribers or fewer
- for each subscriber above 100, the charge to **every** subscriber is reduced by $0.10$

For example, with 110 subscribers, the charge per subscriber is

$$
20 - 10(0.10) = 19 \text{ dollars}.
$$

We want the number of subscribers that gives the **greatest total revenue**.

## Step 1: Write the price per subscriber

Let $n$ be the number of subscribers.

For $n > 100$, the excess above 100 is $n - 100$, so the reduction in price is

$$
0.10(n - 100).
$$

So the price per subscriber is

$$
p(n) = 20 - 0.10(n - 100).
$$

Simplifying,

$$
p(n) = 30 - 0.1n.
$$

## Step 2: Write the revenue function

Revenue equals

$$
\text{revenue} = \text{price} \times \text{number of subscribers}.
$$

So for $n > 100$,

$$
R(n) = n(30 - 0.1n) = 30n - 0.1n^2.
$$

For $n \le 100$, the revenue is just

$$
R(n) = 20n.
$$

## Step 3: Maximize the revenue for $n > 100$

Differentiate:

$$
R'(n) = 30 - 0.2n.
$$

Set this equal to 0:

$$
30 - 0.2n = 0.
$$

So

$$
0.2n = 30
$$

$$
n = 150.
$$

## Step 4: Check against the first pricing region

If $n \le 100$, the revenue is $20n$, which is largest at $n = 100$:

$$
R(100) = 2000.
$$

At $n = 150$,

$$
p(150) = 30 - 0.1(150) = 15,
$$

so the revenue is

$$
R(150) = 150 \cdot 15 = 2250.
$$

Since $2250 > 2000$, the overall maximum occurs at 150 subscribers.

## Intuition

There are two competing effects:

- getting more subscribers increases revenue
- lowering the price for everyone decreases revenue per subscriber

At first, adding subscribers helps more than the discount hurts. After a certain point, the discount hurts too much. The best choice is exactly where these effects balance.

## Final answer

The company collects the greatest amount when there are

$$
\boxed{150 \text{ subscribers}}
$$

and the maximum collection is

$$
\boxed{2250 \text{ dollars}}.
$$

```python
n = sp.symbols('n', real=True)
revenue_after_100 = 30*n - n**2 / 10
critical_n = sp.solve(sp.diff(revenue_after_100, n), n)[0]

print(f'revenue for n > 100: {revenue_after_100}')
print(f'critical number of subscribers = {critical_n}')
print(f'revenue at 100 subscribers = {20*100}')
print(f'revenue at 150 subscribers = {sp.simplify(revenue_after_100.subs(n, 150))}')
```

# Folded Page Problem

The lower outer corner of a page is folded so that it just reaches the **inner edge** of the page.

We want the width of the part folded over, measured along the bottom edge, when the **crease length is as small as possible**.

## Step 1: Set up coordinates

Let the width of the page be $w$.

Place the page so that:

- the inner edge is the line $x = 0$
- the bottom edge is the line $y = 0$
- the lower outer corner is the point $(w, 0)$

Suppose the corner is folded to a point $(0, h)$ on the inner edge.

## Step 2: Use the crease geometry

When a point is folded onto another point, the crease is the **perpendicular bisector** of the segment joining them.

So the segment joining

$$
(w,0) \quad \text{and} \quad (0,h)
$$

has midpoint

$$
\left(\frac{w}{2}, \frac{h}{2}\right),
$$

and slope

$$
\frac{h - 0}{0 - w} = -\frac{h}{w}.
$$

Therefore the crease has slope

$$
\frac{w}{h}.
$$

## Step 3: Find the fold width along the bottom

Let $x$ be the width folded over along the bottom edge.

That means the crease meets the bottom edge at a point whose horizontal distance from the corner $(w,0)$ is $x$.

Using the perpendicular-bisector relation, this turns out to give

$$
x = \frac{w^2 + h^2}{2w}.
$$

## Step 4: Write the crease length

The crease runs from the bottom edge to the outer edge. Its horizontal run is $x$, and since its slope is $\frac{w}{h}$, its vertical rise is

$$
\frac{w}{h}x.
$$

So the crease length $L$ is

$$
L = \sqrt{x^2 + \left(\frac{w}{h}x\right)^2}
= x\sqrt{1 + \frac{w^2}{h^2}}.
$$

Substitute

$$
x = \frac{w^2 + h^2}{2w}
$$

to get

$$
L = \frac{(w^2 + h^2)^{3/2}}{2wh}.
$$

## Step 5: Minimize the crease length

Since $w$ is constant, minimize

$$
L(h) = \frac{(w^2 + h^2)^{3/2}}{2wh}.
$$

Differentiate with respect to $h$ and set equal to 0. This gives

$$
2h^2 - w^2 = 0,
$$

so

$$
h = \frac{w}{\sqrt{2}}.
$$

## Step 6: Find the folded width

Substitute $h = \frac{w}{\sqrt{2}}$ into

$$
x = \frac{w^2 + h^2}{2w}:
$$

$$
x = \frac{w^2 + \frac{w^2}{2}}{2w} = \frac{3w^2/2}{2w} = \frac{3w}{4}.
$$

## Final answer

The width of the part folded over along the bottom edge should be

$$
\boxed{\frac{3}{4} \text{ of the width of the page}}
$$

when the crease length is least possible.

```python
w, h = sp.symbols('w h', positive=True)
fold_width = (w**2 + h**2) / (2*w)
crease_length = sp.simplify((w**2 + h**2)**sp.Rational(3, 2) / (2*w*h))
critical_h = sp.solve(sp.simplify(sp.diff(crease_length, h) * 2*w*h**2 / sp.sqrt(w**2 + h**2)), h)[0]
optimal_width = sp.simplify(fold_width.subs(h, w / sp.sqrt(2)))

print(f'fold width x = {fold_width}')
print(f'crease length L = {crease_length}')
print(f'optimal h = {sp.simplify(w / sp.sqrt(2))}')
print(f'optimal fold width = {optimal_width}')
print(f'fraction of page width = {sp.simplify(optimal_width / w)}')
```
