# Intuition for Special Relativity from Einstein's 1905 Paper

This notebook is a guided study of Einstein's 1905 paper, **On the Electrodynamics of Moving Bodies**.

The goal is not just to list the results, but to understand the order of Einstein's reasoning:

1. Why classical electrodynamics looked conceptually inconsistent.
2. Why the notion of **simultaneity** had to be redefined.
3. How the two postulates force the Lorentz transformation.
4. How time dilation, length contraction, and velocity addition follow.
5. Why the electrodynamics sections matter, and are not just an afterthought.

The paper is really about removing the need for **absolute rest** and showing that Maxwell's theory can be made consistent with the relativity principle.

## 1. The Problem Einstein Starts With

Einstein begins with a specific asymmetry in classical electrodynamics.

If a magnet moves past a conductor, we say the magnet creates an electric field that drives a current.
If the conductor moves past the magnet, we often describe the effect differently: no electric field in the surrounding space, but an electromotive force appears in the conductor.

Yet the **observable current** depends only on the **relative motion** of magnet and conductor.

So the old theory seemed to distinguish two situations that nature itself does not distinguish.

This is Einstein's starting complaint: the equations and concepts were carrying extra structure, especially the idea of an ether or absolute rest, that did not show up in observable phenomena.

He also points to failed attempts to detect Earth's motion through the ether. This suggests that absolute rest is not physically meaningful.

### The same experiment, written two ways

The two equations sitting in the background are Faraday's law and the Lorentz force law:

$$
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}
$$

and

$$
\mathbf{F} = q\left(\mathbf{E} + \mathbf{u} \times \mathbf{B}\right).
$$

Here $\mathbf{u}$ is the velocity of a charge in the frame you are using.

If we describe the situation in the **rest frame of the magnet**, then the magnetic field is static and, ideally,

$$
\frac{\partial \mathbf{B}}{\partial t} = 0, \qquad \mathbf{E} = 0.
$$

A conductor moving with velocity $\mathbf{u}$ through that field feels a magnetic force on its charges:

$$
\mathbf{F} = q\,\mathbf{u} \times \mathbf{B}.
$$

So the emf is explained as a **magnetic effect on moving charges**.

If instead we describe the same setup in the **rest frame of the conductor**, then the magnet is moving, so the field at the conductor changes with time. Now Faraday's law says:

$$
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t} \neq 0,
$$

so an induced electric field appears and the charges are pushed by

$$
\mathbf{F} = q\mathbf{E}.
$$

So the emf is now explained as an **electric effect from a changing magnetic field**.

The important point is that both descriptions predict the **same measurable current**. The split into "electric cause" versus "magnetic cause" depends on the frame, while the phenomenon does not.

That is exactly what bothered Einstein: if only the relative motion matters, why should the theory treat the two descriptions as fundamentally different?

In modern language, special relativity resolves this because electric and magnetic fields are not separate absolute entities. They are frame-dependent parts of one electromagnetic structure, and very roughly they mix as

$$
\mathbf{E}' \approx \mathbf{E} + \mathbf{v} \times \mathbf{B}
$$

for low relative speeds. Einstein did not start with this final formula. He started with the asymmetry above and asked for a mechanics and electrodynamics in which no preferred state of rest is needed.

## 2. Einstein's Two Postulates

Einstein proposes two principles:

These did not come out of nowhere. A plausible reconstruction is that Einstein was trying to preserve two facts at once:

1. The magnet-conductor experiment suggests that only **relative motion** should matter.
2. Maxwell's equations already single out a wave speed,

$$
c = \frac{1}{\sqrt{\mu_0\varepsilon_0}},
$$

which looked like a universal feature of electromagnetism rather than a property of a moving source.

So the postulates can be read less as arbitrary assumptions and more as the simplest way to keep both the relativity idea and Maxwell's electrodynamics.

### Postulate 1: Principle of Relativity

The laws of physics are the same in every inertial frame.

That means there is no preferred inertial frame whose laws are the 'real' ones.

This extends to electrodynamics a principle that already felt natural in mechanics: uniform motion should not be detectable by purely internal experiments.

### Postulate 2: Constancy of the Speed of Light

Light in vacuum travels with speed $c$ in every inertial frame, independent of the motion of the source.

This is the radical step. Taken literally, it clashes with the old intuition that velocities should add in the Galilean way.

A useful way to read Einstein here is: if Maxwell's equations are to hold in every inertial frame, then the light speed coming from those equations should also come out the same in every inertial frame. Instead of modifying Maxwell to fit Galilean kinematics, Einstein modified the kinematics.

Einstein's insight is that we should keep both postulates and instead revise our ideas of time and simultaneity.

### How he might have reasoned

Very roughly, the logic could have been:

1. Ether drift is not showing up in experiment.
2. The relativity principle seems too deep to apply only to mechanics.
3. Maxwell's theory is too successful to discard lightly.
4. Therefore, the conflict is probably in our assumptions about space and time, not in electromagnetism itself.

That is why the postulates are so compressed. They are the cleanest summary of what Einstein seems to have decided must be kept fixed while rebuilding the concepts of time, simultaneity, and transformation between frames.

## 3. The Real Conceptual Break: What Does 'At the Same Time' Mean?

Einstein's deepest move is not length contraction or time dilation. It is the redefinition of **time at distant places**.

Suppose you have clocks at points $A$ and $B$. How do you know they are synchronized?
Einstein says: send a light signal from $A$ to $B$ and back. If the signal leaves $A$ at time $t_A$, reaches $B$ at time $t_B$, and returns to $A$ at time $t'_A$, then define the clocks to be synchronized when

$$
t_B - t_A = t'_A - t_B
$$

This is not a theorem. It is a definition of simultaneity based on the assumption that light takes the same time in both directions.

Once you adopt that rule, simultaneity becomes frame-dependent. Two events that are simultaneous in one inertial frame need not be simultaneous in another.

This is the conceptual heart of the paper. Absolute time is replaced by frame-dependent time.

## 4. From the Postulates to the Lorentz Transformation

Einstein next asks: if two inertial frames move at constant relative velocity $v$, how must space and time coordinates transform so that the two postulates remain true?

The answer is the Lorentz transformation:

$$
x' = \gamma(x-vt),
\qquad
t' = \gamma\left(t - \frac{vx}{c^2}\right),
$$

$$
y' = y,
\qquad
z' = z,
\qquad
\gamma = \frac{1}{\sqrt{1-v^2/c^2}}
$$

The crucial feature is that **time mixes with space**. In Galilean mechanics, time is unchanged: $t' = t$. In special relativity, time depends on both $t$ and $x$.

That single change explains why simultaneity becomes relative.

## 5. Immediate Consequences

From the Lorentz transformation, Einstein extracts several consequences.

### Relativity of simultaneity

If two events satisfy $t_1 = t_2$ in one frame but happen at different positions $x_1 \neq x_2$, then in another frame moving with speed $v$,

$$
t'_2 - t'_1 = \gamma\left[-\frac{v(x_2-x_1)}{c^2}\right]
$$

so they are generally **not simultaneous**.

### Time dilation

A moving clock runs slow relative to a stationary observer:

$$
\Delta t = \gamma\,\Delta \tau
$$

where $\Delta \tau$ is the proper time measured by the clock itself.

### Length contraction

A rod of proper length $L_0$ moving parallel to its length is measured as

$$
L = \frac{L_0}{\gamma}
$$

by an observer relative to whom the rod is moving.

### Velocity addition

Velocities do not add as $u+v$. Instead, for motion along one axis,

$$
u' = \frac{u-v}{1-uv/c^2}
$$

This ensures that no material object is boosted past the speed of light.

## 6. Why the Electrodynamics Part Matters

Many introductions stop after time dilation and length contraction. Einstein's paper does not.

The second half of the paper shows that Maxwell's equations keep their form under the new transformations, and that electric and magnetic fields transform into one another between frames.

This resolves the magnet-conductor asymmetry from the introduction. What looked like two physically different situations in the old language are really two descriptions of the same electromagnetic process in different inertial frames.

A modern summary is:

- electric and magnetic fields are frame-dependent aspects of one electromagnetic structure,
- there is no need for an ether frame,
- the laws of electrodynamics are compatible with the relativity principle.

This is why the paper is called **On the Electrodynamics of Moving Bodies** rather than simply **On Time and Space**.

## 7. Section-by-Section Map of the 1905 Paper

Here is a useful way to remember the paper's internal structure:

1. **Introduction**: identify the conceptual asymmetry in electrodynamics and reject absolute rest.
2. **Synchronism**: define time operationally using light signals.
3. **Relativity of time and length**: show that simultaneity is frame-dependent.
4. **Coordinate transformation**: derive the Lorentz transformation.
5. **Physical meaning**: explain moving rods, clocks, and clock transport.
6. **Velocity addition**: show that $c$ remains the limiting speed.
7. **Field transformations**: show how electric and magnetic fields transform.
8. **Optics of moving bodies**: derive Doppler shift, aberration, and radiation-pressure results.
9. **Charged matter and currents**: transform Maxwell's equations with sources.
10. **Electron dynamics**: derive consequences for accelerated charged particles.

This is the flow of the original paper: conceptual repair first, mathematics second, electrodynamics third.

## 8. What to Keep in Mind While Studying

The most important lessons of the paper are:

- Einstein does **not** start with time dilation. He starts with the meaning of simultaneity.
- The Lorentz transformation is not an arbitrary trick. It is forced by the two postulates.
- Special relativity is not only kinematics. It is also a reformulation of electrodynamics.
- Electric and magnetic fields are not absolute separate objects; they mix under changes of frame.
- The speed of light is not just one more speed. It is the invariant conversion factor between space and time in the theory.

If you understand those five ideas, you understand the essence of the paper.
