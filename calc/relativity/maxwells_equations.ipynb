# Maxwell's Equations

Maxwell's equations summarize classical electromagnetism. They describe how electric fields $\mathbf{E}$ and magnetic fields $\mathbf{B}$ are produced by charges and currents, and how they evolve in space and time.

## 1. Gauss's Law for Electricity

### Differential form

$$
\nabla \cdot \mathbf{E} = \frac{\rho}{\varepsilon_0}
$$

### Integral form

$$
\oint_S \mathbf{E} \cdot d\mathbf{A} = \frac{Q_{\text{enc}}}{\varepsilon_0}
$$

### Meaning

Electric charge is the source of electric field. A positive charge acts like a source from which electric field lines emerge, and a negative charge acts like a sink into which they terminate.

## 2. Gauss's Law for Magnetism

### Differential form

$$
\nabla \cdot \mathbf{B} = 0
$$

### Integral form

$$
\oint_S \mathbf{B} \cdot d\mathbf{A} = 0
$$

### Meaning

There are no isolated magnetic charges, at least in ordinary classical electromagnetism. Magnetic field lines do not begin or end at a point; they form closed loops.

## 3. Faraday's Law of Induction

### Differential form

$$
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}
$$

### Integral form

$$
\oint_C \mathbf{E} \cdot d\mathbf{l} = -\frac{d}{dt}\int_S \mathbf{B} \cdot d\mathbf{A}
$$

### Meaning

A changing magnetic field produces a circulating electric field. This is the basic principle behind electromagnetic induction, transformers, and electric generators.

## 4. Ampere-Maxwell Law

### Differential form

$$
\nabla \times \mathbf{B} = \mu_0 \mathbf{J} + \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}
$$

### Integral form

$$
\oint_C \mathbf{B} \cdot d\mathbf{l} = \mu_0 I_{\text{enc}} + \mu_0 \varepsilon_0 \frac{d}{dt}\int_S \mathbf{E} \cdot d\mathbf{A}
$$

### Meaning

Magnetic fields are produced both by electric currents and by changing electric fields. Maxwell's extra term,

$$
\mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t},
$$

is the displacement current term. It is what allows electromagnetic waves to exist.

## Maxwell's Equations in Vacuum

In vacuum, there are no charges and no currents, so

$$
\rho = 0, \qquad \mathbf{J} = 0.
$$

The equations become

$$
\nabla \cdot \mathbf{E} = 0, \qquad \nabla \cdot \mathbf{B} = 0,
$$

$$
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t},
$$

$$
\nabla \times \mathbf{B} = \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}.
$$

These are enough to derive the wave equation.

## Deriving the Speed of Light

Start from Faraday's law:

$$
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}.
$$

Take the curl of both sides:

$$
\nabla \times (\nabla \times \mathbf{E}) = -\frac{\partial}{\partial t}(\nabla \times \mathbf{B}).
$$

Now use the Ampere-Maxwell law in vacuum:

$$
\nabla \times \mathbf{B} = \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}.
$$

Substituting gives

$$
\nabla \times (\nabla \times \mathbf{E}) = -\mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{E}}{\partial t^2}.
$$

Next use the vector identity

$$
\nabla \times (\nabla \times \mathbf{E}) = \nabla(\nabla \cdot \mathbf{E}) - \nabla^2 \mathbf{E}.
$$

In vacuum, Gauss's law gives

$$
\nabla \cdot \mathbf{E} = 0,
$$

so this becomes

$$
-\nabla^2 \mathbf{E} = -\mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{E}}{\partial t^2}.
$$

Therefore,

$$
\nabla^2 \mathbf{E} = \mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{E}}{\partial t^2}.
$$

This is a wave equation. Compare it with the standard wave equation

$$
\nabla^2 \psi = \frac{1}{v^2} \frac{\partial^2 \psi}{\partial t^2},
$$

and you can read off

$$
\frac{1}{v^2} = \mu_0 \varepsilon_0.
$$

So the wave speed is

$$
v = \frac{1}{\sqrt{\mu_0 \varepsilon_0}}.
$$

That speed matches the measured speed of light, so Maxwell's theory predicts that light is an electromagnetic wave.

You can do the same derivation for $\mathbf{B}$ and get

$$
\nabla^2 \mathbf{B} = \mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{B}}{\partial t^2}.
$$

## Why These Equations Matter

Taken together, these equations imply that electric and magnetic fields can sustain one another and propagate through empty space as waves. From them one finds the wave speed

$$
c = \frac{1}{\sqrt{\mu_0 \varepsilon_0}},
$$

which is the speed of light in vacuum.

That is one of the main reasons Maxwell's equations are so central to the story of special relativity: they already contain a distinguished speed, and Einstein asked how the laws of physics must be organized if that speed is the same in all inertial frames.

## Deriving the Magnetic-Field Wave Equation

There is a completely symmetric derivation for the magnetic field.

Start from the Ampere-Maxwell law in vacuum:

$$
\nabla \times \mathbf{B} = \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}.
$$

Take the curl of both sides:

$$
\nabla \times (\nabla \times \mathbf{B}) = \mu_0 \varepsilon_0 \frac{\partial}{\partial t}(\nabla \times \mathbf{E}).
$$

Now use Faraday's law:

$$
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}.
$$

Substituting gives

$$
\nabla \times (\nabla \times \mathbf{B}) = -\mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{B}}{\partial t^2}.
$$

Next use the vector identity

$$
\nabla \times (\nabla \times \mathbf{B}) = \nabla(\nabla \cdot \mathbf{B}) - \nabla^2 \mathbf{B}.
$$

In vacuum, Gauss's law for magnetism gives

$$
\nabla \cdot \mathbf{B} = 0,
$$

so this becomes

$$
-\nabla^2 \mathbf{B} = -\mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{B}}{\partial t^2}.
$$

Therefore,

$$
\nabla^2 \mathbf{B} = \mu_0 \varepsilon_0 \frac{\partial^2 \mathbf{B}}{\partial t^2}.
$$

This has the same wave speed as the electric-field equation:

$$
v = \frac{1}{\sqrt{\mu_0 \varepsilon_0}}.
$$

So in vacuum, both $\mathbf{E}$ and $\mathbf{B}$ propagate together as electromagnetic waves at the speed of light.
