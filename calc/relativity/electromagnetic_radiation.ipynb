# Electromagnetic Radiation

Electromagnetic radiation is a self-propagating disturbance of electric and magnetic fields.

The key idea is this:

- a changing electric field creates a magnetic field
- a changing magnetic field creates an electric field
- together they can travel through empty space as a wave

Light, radio waves, microwaves, infrared, ultraviolet, X-rays, and gamma rays are all forms of electromagnetic radiation. They differ mainly by wavelength and frequency.

## Core picture

In a simple plane electromagnetic wave moving in the $x$-direction:

- the electric field $\mathbf{E}$ oscillates in one transverse direction
- the magnetic field $\mathbf{B}$ oscillates in a perpendicular transverse direction
- the wave travels perpendicular to both of them

A standard picture is:

$$
\mathbf{E} \perp \mathbf{B} \perp \text{direction of travel}
$$

For a wave in vacuum, the fields oscillate in phase and satisfy

$$
E = cB,
$$

where $c$ is the speed of light.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 4 * np.pi, 600)
electric = np.sin(x)
magnetic = 0.8 * np.sin(x)

fig, ax = plt.subplots(figsize=(11, 4.5))
ax.plot(x, electric, label='Electric field E', color='royalblue', linewidth=2.5)
ax.plot(x, magnetic, label='Magnetic field B', color='darkorange', linewidth=2.5)
ax.axhline(0, color='0.7', linewidth=1)
ax.set_title('Electric and magnetic fields oscillate together')
ax.set_xlabel('Position along direction of travel')
ax.set_ylabel('Field strength (scaled)')
ax.grid(alpha=0.25)
ax.legend()
plt.show()
```

## Why Maxwell's equations predict radiation

In empty space, Maxwell's equations reduce to:

$$
\nabla \cdot \mathbf{E} = 0, \qquad \nabla \cdot \mathbf{B} = 0,
$$

$$
\nabla \times \mathbf{E} = -\frac{\partial \mathbf{B}}{\partial t}, \qquad \nabla \times \mathbf{B} = \mu_0 \varepsilon_0 \frac{\partial \mathbf{E}}{\partial t}.
$$

These imply wave equations for both $\mathbf{E}$ and $\mathbf{B}$, so the fields can propagate through vacuum with speed

$$
c = \frac{1}{\sqrt{\mu_0 \varepsilon_0}}.
$$

That speed is the speed of light, which is why light is an electromagnetic wave.

## Frequency, wavelength, and energy

All electromagnetic radiation travels at the same speed in vacuum:

$$
c = f\lambda
$$

where:

- $f$ is frequency
- $\lambda$ is wavelength
- $c$ is the speed of light

Higher frequency means shorter wavelength. In quantum physics, the photon energy is

$$
E = hf,
$$

so higher-frequency radiation also carries more energy per photon.

```python
bands = [
    ('Radio', 1e5, 1e9, '#4c78a8'),
    ('Microwave', 1e9, 3e11, '#72b7b2'),
    ('Infrared', 3e11, 4e14, '#f58518'),
    ('Visible', 4e14, 7.5e14, '#e45756'),
    ('Ultraviolet', 7.5e14, 3e16, '#54a24b'),
    ('X-ray', 3e16, 3e19, '#b279a2'),
    ('Gamma', 3e19, 1e22, '#9d755d'),
]

fig, ax = plt.subplots(figsize=(12, 2.8))
for label, left, right, color in bands:
    ax.barh(0, right - left, left=left, height=0.55, color=color, edgecolor='none')
    ax.text(np.sqrt(left * right), 0, label, ha='center', va='center', fontsize=10, color='white')

ax.set_xscale('log')
ax.set_yticks([])
ax.set_xlabel('Frequency (Hz)')
ax.set_title('Electromagnetic spectrum by frequency')
ax.spines['left'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
plt.show()
```

## Examples across the spectrum

- **Radio waves**: communication, broadcasting, astronomy
- **Microwaves**: radar, Wi-Fi, cooking
- **Infrared**: heat radiation, thermal imaging
- **Visible light**: human vision
- **Ultraviolet**: fluorescence, sunburn, sterilization
- **X-rays**: medical imaging, material inspection
- **Gamma rays**: nuclear processes, high-energy astrophysics

The physics is unified: these are all the same kind of wave, just at different frequencies and wavelengths.

## Important ideas to remember

1. Electromagnetic radiation does not need a material medium to travel.
2. Electric and magnetic fields sustain each other as the wave moves.
3. The wave is transverse: the fields oscillate perpendicular to the direction of travel.
4. All electromagnetic waves move at speed $c$ in vacuum.
5. Frequency, wavelength, and energy are tightly linked.

A clean mental model is: electromagnetic radiation is a traveling pattern of mutually linked electric and magnetic oscillations.
