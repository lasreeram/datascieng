# Falling Ball Motion Analysis

This notebook visualizes the three fundamental equations of motion for a ball falling from a certain height:
- Acceleration: $a(t) = g$
- Velocity: $v(t) = gt + v_0$
- Position: $s(t) = \frac{1}{2}gt^2 + v_0t + s_0$

The origin is at the starting point, with downward direction as positive.

## Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Define Physical Parameters

```python
# Physical constants and initial conditions
g = 9.8  # gravitational acceleration (m/s^2), positive downward
v0 = 0   # initial velocity (m/s), starting from rest
s0 = 0   # initial position (m), origin at starting point

# Time range for simulation
t = np.linspace(0, 3, 300)  # 0 to 3 seconds
```

## Calculate Motion Equations

```python
# Calculate acceleration, velocity, and position over time
acceleration = np.full_like(t, g)  # a(t) = g (constant)
velocity = g * t + v0              # v(t) = gt + v0
position = 0.5 * g * t**2 + v0 * t + s0  # s(t) = (1/2)gt^2 + v0*t + s0
```

## Plot the Three Graphs

```python

# Create figure with three subplots
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 10))

# Plot 1: Acceleration vs Time
ax1.plot(t, acceleration, 'r-', linewidth=2)
ax1.set_xlabel('Time (s)')
ax1.set_ylabel('Acceleration (m/s²)')
ax1.set_title('Acceleration: a(t) = g')
ax1.grid(True, alpha=0.3)
ax1.axhline(y=0, color='k', linestyle='-', linewidth=0.5)

# Plot 2: Velocity vs Time
ax2.plot(t, velocity, 'b-', linewidth=2)
ax2.set_xlabel('Time (s)')
ax2.set_ylabel('Velocity (m/s)')
ax2.set_title('Velocity: v(t) = gt + v₀')
ax2.grid(True, alpha=0.3)
ax2.axhline(y=0, color='k', linestyle='-', linewidth=0.5)

# Plot 3: Position vs Time
ax3.plot(t, position, 'g-', linewidth=2)
ax3.set_xlabel('Time (s)')
ax3.set_ylabel('Position (m)')
ax3.set_title('Position: s(t) = ½gt² + v₀t + s₀')
ax3.grid(True, alpha=0.3)
ax3.axhline(y=0, color='k', linestyle='-', linewidth=0.5)

plt.tight_layout()
plt.show()
```
