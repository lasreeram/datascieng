# Circular Plate Heat Equation in Polar Coordinates

## Why polar coordinates are the natural language here

For a circular plate, the geometry itself is radial: every point is described naturally by

- its distance from the center, $r$
- its angle around the center, $\theta$

So instead of working with $(x,y)$, it is much more natural to work with $(r,\theta)$.

If the temperature is in steady state, then there is no time dependence and the temperature $u(r,\theta)$ satisfies Laplace's equation. In polar coordinates that becomes:

$$u_{rr} + \frac{1}{r}u_r + \frac{1}{r^2}u_{\theta\theta} = 0$$

The extra terms are not arbitrary. They reflect the geometry of circles:

- moving outward changes the circumference available for heat flow
- angular variation matters less near the center because points with different angles are physically very close together

That is exactly why the factor $\dfrac{1}{r^2}$ multiplies the angular derivative term.

## The main physical intuition

Heat wants to smooth out differences. On a circular plate that smoothing happens in two directions:

- radially: from the boundary toward the center
- angularly: around each circular ring

A useful way to think about the solution is to decompose the boundary temperature into angular patterns.

If the boundary at $r=R$ has temperature profile $f(\theta)$, then each angular pattern propagates inward differently.

- the average temperature mode survives everywhere
- low-frequency angular patterns penetrate deep into the disk
- rapid oscillations in angle die out quickly as you move inward

So the center of the disk only feels the broadest features of the boundary data. Fine angular wiggles near the edge are rapidly damped away.

---

## Separation of variables in polar form

Assume a product solution

$$u(r,\theta)=R(r)\Theta(\theta)$$

Substituting into Laplace's equation gives radial and angular ODEs. The angular part satisfies

$$\Theta'' + n^2\Theta = 0$$

so the angular eigenfunctions are

$$\cos(n\theta), \qquad \sin(n\theta), \qquad n=0,1,2,\dots$$

These are exactly the natural angular wave patterns around a circle.

The radial part becomes

$$r^2R'' + rR' - n^2R = 0$$

whose solutions are

$$R(r)=Ar^n + Br^{-n}$$

For a solid disk that includes the center, we reject the $r^{-n}$ term because it blows up at $r=0$. So only the regular part remains:

$$R(r) \propto r^n$$

That gives the interior steady-state solution

$$u(r,\theta)=a_0 + \sum_{n=1}^{\infty} r^n\big(a_n\cos(n\theta)+b_n\sin(n\theta)\big)$$

If the boundary is imposed at radius $R$, it is cleaner to write

$$u(r,\theta)=a_0 + \sum_{n=1}^{\infty} \left(\frac{r}{R}\right)^n\big(a_n\cos(n\theta)+b_n\sin(n\theta)\big)$$

so each mode is normalized to the boundary circle.

---

## What the factor $\left(\frac{r}{R}\right)^n$ means

This is the key intuition for the circular plate:

- $n=0$: the average mode does not decay; it fills the whole disk
- $n=1$: a broad warm-to-cool gradient reaches the center easily
- $n=2$: variation decays faster
- large $n$: sharp angular oscillations are almost completely confined to the outer rim

So higher angular frequency means faster radial decay. The disk acts like a spatial low-pass filter.

```python
import numpy as np
import matplotlib.pyplot as plt

R = 1.0
N = 300
x = np.linspace(-R, R, N)
y = np.linspace(-R, R, N)
X, Y = np.meshgrid(x, y)
r = np.sqrt(X**2 + Y**2)
theta = np.arctan2(Y, X)

mask = r <= R

def disk_mode(n, kind="cos"):
    if n == 0:
        U = np.ones_like(r)
    else:
        angular = np.cos(n * theta) if kind == "cos" else np.sin(n * theta)
        U = (r / R) ** n * angular
    U = np.where(mask, U, np.nan)
    return U

modes = [(0, "cos"), (1, "cos"), (4, "cos")]
titles = [r"$n=0$ (average mode)", r"$n=1$ (broad variation)", r"$n=4$ (rapid angular wiggles)"]

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), constrained_layout=True)

for ax, (n, kind), title in zip(axes, modes, titles):
    U = disk_mode(n, kind)
    im = ax.contourf(X, Y, U, levels=30, cmap="coolwarm")
    ax.contour(X, Y, U, levels=10, colors="black", linewidths=0.4, alpha=0.4)
    boundary_x = R * np.cos(np.linspace(0, 2 * np.pi, 400))
    boundary_y = R * np.sin(np.linspace(0, 2 * np.pi, 400))
    ax.plot(boundary_x, boundary_y, color="black", linewidth=1.0)
    ax.set_title(title)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

fig.colorbar(im, ax=axes, shrink=0.85)
plt.show()
```

## How to read the plots

Each panel shows one pure angular boundary mode continued into the interior of the disk.

- Left: $n=0$ is constant everywhere. This is the average boundary temperature.
- Middle: $n=1$ produces one warm side and one cool side, and that broad pattern reaches the center.
- Right: $n=4$ oscillates rapidly around the boundary, but it fades strongly before reaching the center.

This is the most important intuition for the circular steady-state heat problem:

$$\text{higher angular complexity on the boundary dies out faster toward the center}$$

So if you know the boundary temperature $f(\theta)$, the interior temperature is just the sum of these modes, each multiplied by its radial decay factor $\left(\dfrac{r}{R}\right)^n$.

---

## One more geometric insight

Near the center, all angles are squeezed together into a tiny physical region. That makes it hard for strong angular temperature differences to persist there.

So the center mostly remembers the average boundary temperature, while detailed angular structure lives near the outer edge.

If you later study an annulus instead of a solid disk, the discarded $r^{-n}$ terms become admissible because there is no center where they blow up. That changes the radial behavior substantially.
